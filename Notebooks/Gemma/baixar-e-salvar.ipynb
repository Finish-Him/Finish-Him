# 📥 Baixar Arquivos BAML e Salvar no Google Drive

Este notebook:
1. Conecta ao Google Drive
2. Baixa arquivos do GitHub (Finish-Him/Finish-Him)
3. Salva na pasta BAML

## Arquivos que serão baixados:
- gemma-3-4b-chat.ipynb
- gemma-1b-chat.ipynb
- salvar-no-drive.ipynb

```
# 1. Conectar Google Drive
from google.colab import drive
drive.mount('/content/drive')
print('✅ Drive conectado!')
```

```
# 2. Criar pasta BAML no Drive
import os
baml_path = '/content/drive/My Drive/BAML'
os.makedirs(baml_path, exist_ok=True)
print(f'📁 Pasta criada: {baml_path}')
```

```
# 3. Clonar repositório
!git clone https://github.com/Finish-Him/Finish-Him.git /tmp/repo
print('✅ Repositório clonado!')
```

```
# 4. Copiar arquivos
import shutil

files = [
    '/tmp/repo/gemma-3-4b-chat/gemma-3-4b-chat.ipynb',
    '/tmp/repo/gemma-3-4b-chat/gemma-1b-chat.ipynb',
    '/tmp/repo/gemma-3-4b-chat/salvar-no-drive.ipynb'
]

for f in files:
    dest = f'{baml_path}/{os.path.basename(f)}'
    if os.path.exists(f):
        shutil.copy(f, dest)
        print(f'✅ {os.path.basename(f)}')
    else:
        print(f'❌ Não encontrado: {f}')
```

```
# 5. Listar arquivos salvos
print('\n📁 Arquivos no Drive:')
for f in os.listdir(baml_path):
    print(f'  - {f}')
```

## ✅ Concluído!

Seus arquivos estão em: `/content/drive/My Drive/BAML/`
