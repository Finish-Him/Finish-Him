```
# Conectar Drive
from google.colab import drive
drive.mount('/content/drive')
print('✅ Drive conectado!')
```

```
import os

# Pasta de destino
from pydrive2.auth import GoogleAuth
from pydrive2.drive import GoogleDrive

gauth = GoogleAuth()
gauth.CommandLineAuth()
drive = GoogleDrive(gauth)

folder_id = '1zhorJ03BIV3T5C-HlJVRaFGwLfe8vl2p'

print('🔗 Conectado ao Drive!')
```

```
# Arquivos para upload
files = [
    '/root/workspace/BAML/Colab/gemma-3-4b-chat.ipynb',
    '/root/workspace/BAML/Colab/gemma-1b-chat.ipynb',
    '/root/workspace/BAML/Colab/salvar-no-drive.ipynb'
]

for file_path in files:
    if os.path.exists(file_path):
        file_name = os.path.basename(file_path)
        
        # Criar arquivo no Drive
        f = drive.CreateFile({
            'title': file_name,
            'parents': [{'id': folder_id}]
        })
        f.SetContentFile(file_path)
        f.Upload()
        print(f'✅ {file_name}')
```
