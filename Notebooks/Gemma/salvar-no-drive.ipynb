# 💾 Salvar Arquivos BAML no Google Drive

Este notebook copia os arquivos BAML para seu Google Drive.

## Arquivos que serão salvos:
- gemma-3-4b-chat.ipynb (versão 4B)
- gemma-1b-chat.ipynb (versão 1B)
- vps-gemma-server.sh (script VPS)
- README.md (documentação)

```
# Conectar Google Drive
from google.colab import drive
drive.mount('/content/drive')

print('✅ Drive conectado!')
```

```
import os
import shutil

# Criar pasta
os.makedirs('/content/drive/My Drive/BAML', exist_ok=True)
print('📁 Pasta criada: /content/drive/My Drive/BAML')
```

```
# Definir arquivos
base_path = '/root/workspace/BAML/Colab'
files = [
    'gemma-3-4b-chat.ipynb',
    'gemma-1b-chat.ipynb',
    'vps-gemma-server.sh',
    'README.md'
]

# Copiar arquivos
for f in files:
    src = f'{base_path}/{f}'
    dest = f'/content/drive/My Drive/BAML/{f}'
    
    if os.path.exists(src):
        shutil.copy(src, dest)
        print(f'✅ {f}')
    else:
        print(f'⚠️ {f} - não encontrado')
```

## ✅ Concluído!

Seus arquivos estão em: `/content/drive/My Drive/BAML/`
