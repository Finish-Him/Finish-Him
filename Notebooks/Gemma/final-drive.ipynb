```
# Criar pasta e copiar arquivos
import shutil
import os

dest = '/content/drive/My Drive/BAML'
os.makedirs(dest, exist_ok=True)

files = ['gemma-3-4b-chat.ipynb', 'gemma-1b-chat.ipynb']
for f in files:
    src = f'/tmp/gemma-chat/{f}'
    if os.path.exists(src):
        shutil.copy(src, dest)
        print(f'✅ {f} salvo!')
    else:
        print(f'⚠️ {f} não encontrado')
```
