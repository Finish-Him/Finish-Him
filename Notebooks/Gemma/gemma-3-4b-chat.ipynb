2. Aguarde o modelo carregar (~5 minutos)

```python
# 2. Carregar modelo
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

MODEL_NAME = "google/gemma-3-4b-it"

print("📥 Carregando tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("📥 Carregando modelo (~5 minutos)...")
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float32,
    device_map="cpu"
)

print("✅ Modelo carregado!")

# Criar pipeline
pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=512,
    temperature=0.7,
    do_sample=True
)
```

```python
# 3. Interface Gradio
import gradio as gr

def chat(message, history):
    prompt = f"<start_of_turn>user\n{message}<end_of_turn>\n<start_of_turn>model\n"
    output = pipe(prompt)[0]["generated_text"]
    
    # Extrair resposta
    if "<start_of_turn>model\n" in output:
        response = output.split("<start_of_turn>model\n")[-1]
    else:
        response = output
    
    return response.strip()

# Interface
demo = gr.ChatInterface(
    fn=chat,
    title="💬 Gemma 3 4B Chat (Colab - Gratuito)",
    description="Rode Gemma 3 4B de graça no Google Colab!\n⚠️ Mantenha esta aba aberta.",
    examples=[
        ["Ola, tudo bem?"],
        ["Me conte uma piada"],
        ["O que é machine learning?"],
        ["Escreva um poema sobre tecnologia"]
    ]
)

print("🚀 Iniciando interface...")
demo.launch(share=True, server_port=7860)
```
